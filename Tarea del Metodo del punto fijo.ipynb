#Iteracion de punto fijo

El **Método del punto fijo** se basa en el uso del teorema del mismo nombre, el cual nos dice lo siguiente:

Sea $g \in C[a,b] \mid g(x) \in [a,b] \forall x \in [a,b]$ y, suponiendo que $\exists g'(x) \in [a,b]$ y $K≤1 \mid \lvert g'(x) \rvert≤K<1$, entonces para cualquier $P_0 \in [a,b]$ la sucesión definida por $P_n=g(P_{n-1})$ con $n≥1$ converge en el **único** *punto fijo*  $p \in [a,b]$

Por lo tanto, si $f(x)=0 \rightarrow \exists g_i(x)=x-f(x)=x$, justo aquí es donde tenemos un ejemplo de punto fijo en una función, que nos lo garantiza el teorema anterior, también puede haber casos en los que no se cumpla el teorema y aún así funcione el método del punto fijo, pero si se cumple el teorema siempre se cumplirá.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3+4*x**2-10
```

Para el ejemplo tenemos la función $g(x)=x^3+4x^2-10=0$ y dado que cumple el teorema anterior esta función, solo tenemos que despejar una variable $x$ de la ecuación, utilizaremos para este ejemplo el siguiente despeje:$$x= \sqrt{\frac{10}{4+x}}$$

```python
xx=np.linspace(-1,1)
plt.plot(xx,f(xx))
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
```

```python
def g(x):
    return np.sqrt(10/(4+x)) #definimos la función despejada a utilizar para las iteraciones del metodo
```

Es aquí cuando utilizamos el despeje antes mencionado y comenzamos la iteración del método con el punto $P_0=1.5$

```python
def pfijo(p0,i): #definimos una función que englobe todo el algoritmo

    lista=[] #lista donde se guardarán los valores de p1
    error=[] #lista donde se guardarán los valores del error absoluto según vaya cambiando el p1 y p0

    for i in range(i):#El for se repetirá i veces (esperando que sea finito el algoritmo) asignando a p1 el valor de g(p0) y rotando los valores entre p1 y p0
        p1=g(p0)#Se calcula p1
        lista.append(p1)#va guardando los valores de p1 en la lista "lista"
        e=np.abs(p1-p0)#Se calcula e
        error.append(e)#Se guarda el valor de e en la lista
        p0=p1#Se reasigna el p1 a p0 y se repite el for
    return lista, error

p0 = 1.5
n = 10
lista, error = pfijo(p0, n)
p = [p0] + lista

print("{0} \t\t {1} \t\t {2} \t\t {3}".format("n", "Pn" , "g(Pn)", "Error"))
print("{0} \t\t {1:6.6f} \t\t {2:6.6f} \t ----".format(0, p[0], g(p[0])))

for n in range(1, len(p)):
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f}".format(n, p[n], g(p[n]), error[n-1]))

print("La solución aproximadamente es p0 = {0:6.4f} es {1:6.7f}".format(p0, lista[-1]))
```
