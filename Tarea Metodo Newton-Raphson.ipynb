#Metodo de Newton-Raphson

El método de **Newton-Raphson** consiste en el uso de los polinomios de *Taylor*.
Supongamos que $f \in C[a,b]$. Sea $x' \in [a,b]$ una aproximación tal que $f'(x')≠0$ y $ \lvert p - x' \rvert $ "pequeño". Consideremos el primer polinomio de *Taylor* $f(x)$ expendido alrededor de $x'$ $$f(x)=f(x')+(p-x')f'(x')+\frac{(x-x')^2}{2}f''(g(x))$$ Desarrollando está expresión llegamos a la siguiente: $$p=x'-\frac{f(x')}{f'(x')}$$ y haciendo $x'=p_{-1}$ tenemos que: $$p_n=p_{-1}-\frac{f(p_{-1})}{f'(p_{-1})}$$ para $n≥1$

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import sympy as sp
```

```python
def f(x):
    return sp.cos(x)
```

Para nuestro ejercicio, ocuparemos la función $g(x)=cos(x)$, por lo que nuestra función a utilizar en el método será: $$p=p_0-\frac{cos(p_0)-p0}{-sen(p0)-1}$$y le asignaremos un $p_0=\frac{π}{4}$

```python
xx=np.linspace(0,1)
plt.plot(xx,f(xx))
```

```python
def Newton(p0,tol):
    e=100
    p0=float(p0)
    error=[]
    lista=[]
    while e>tol:
      p1=p0-(np.cos(p0)-p0)/(-np.sin(p0)-1)
      lista.append(p1)
      e=np.abs(p1-p0)
      error.append(e)
      p0=p1
    return lista, error
lista, error = Newton(np.pi/4, 0.000001)
print("El valor de P al converger es: ", lista[-1])
```
