#Metodo de Neville
Utilizaremos el método de *Neville* para *aproximar* ($interpolar$) la función desconocida en un punto a partir de polinomios de $0$ hasta $4$.

```python
import numpy as np
import pandas as pd
```

Ingresamos los valores de los $nodos$ y los de los $polinomios$ de grado $0$ para comenzar el algoritmo.

```python
#nodos
x_values = np.array([1.0, 1.3, 1.6, 1.9, 2.2])

y_values = np.array([
    0.7651977,
    0.6200860,
    0.4554022,
    0.2818186,
    0.1103623
])

#Punto a aproximar:
x = 1.5
```

Utilizaremos el algoritmo dado por la expresión: $$Q=\frac{(x-x_{i-j})Q_{i,j-1}-(x-x_i)Q_{i-1,j-1}}{x_i-x_{i-j}}$$
El algoritmo comienza a partir de los polinomios de grado $0$ que se nos dan al inicio.

```python
#definimos la función
def neville(x, x_values, y_values):

    n = len(x_values)

    #Creamos matriz Q utilizando
    Q = np.zeros((n, n))

    Q[:, 0] = y_values

    #Se itera el método cambiando los valores con respecto a j e i
    for j in range(1, n):

        for i in range(j, n):

            Q[i, j] = (
                (x - x_values[i-j]) * Q[i, j-1]
                - (x - x_values[i]) * Q[i-1, j-1]
            ) / (x_values[i] - x_values[i-j])

    return Q
```

Armamos el arreglo para ir mostrando los valores de los $nodos$ y $polinomios$.

```python
Q = neville(x, x_values, y_values)

#Nombres de columna
columnas = [
    f"Q{i}" for i in range(len(x_values))
]

#creamos la tabla con pandas
tabla = pd.DataFrame(
    Q,
    columns=columnas,
    index=x_values
)

#Se muestra la tabla de valores
tabla.index.name = "x_i"

tabla = tabla.where(
    np.tril(np.ones(Q.shape, dtype=bool))
)

#Se muestran los resultados
print("TABLA DE NEVILLE")
display(tabla.round(7))
```

Al final tenemos la aproximación con el polinomio de grado $4$.

```python
resultado = Q[-1, -1]

print("RESULTADO FINAL")
print(f"f(1.5) ≈ {resultado:.7f}")
```
