#Método de diferencias divididas de Newton

El método permite construir un polinomio interpolante a partir de $n+1$ puntos distintos.

**Expresión del algoritmo:**
$$f[x_i]=f(x_i)$$

$$f[x_i,\ldots,x_{i+j}]=\frac{f[x_{i+1},\ldots,x_{i+j}]f[x_i,\ldots,x_{i+j-1}]}{x_{i+j}-x_i}$$

**Polinomio interpolante:**

$$
P_n(x)=\sum_{j=0}^{n}f[x_0,\ldots,x_j]
\prod_{k=0}^{j-1}(x-x_k)
$$

```python
import numpy as np
import pandas as pd
import scipy
```

```python
# Datos del ejemplo 1, página 94

x = np.array([1.0, 1.3, 1.6, 1.9, 2.2])

y = np.array([
    0.7651977,
    0.6200860,
    0.4554022,
    0.2818186,
    0.1103623
])

n = len(x)
```

## 2. Datos iniciales

Se utilizan los cinco puntos del ejemplo 1 de la página 94:
$$x=(1.0,1.3,1.6,1.9,2.2)$$

$$f(x)=(0.7651977,0.6200860,0.4554022,0.2818186,0.1103623)$$

Al contar con cinco puntos, se construye un polinomio de grado máximo cuatro.

```python
def diferencias_divididas(x, y):

    n = len(x)

    # Crear matriz
    D = np.zeros((n, n))

    # Primera columna
    D[:, 0] = y

    # Calcular diferencias divididas
    for j in range(1, n):

        for i in range(n - j):

            D[i, j] = (
                D[i + 1, j - 1] - D[i, j - 1]
            ) / (x[i + j] - x[i])

    return D


D = diferencias_divididas(x, y)
```

## 3. Organización de las diferencias divididas

Se crea una matriz $D$ para almacenar las diferencias de distintos órdenes.

$$
D_{i,0}=f(x_i)
$$

$$
D_{i,j}=
\frac{D_{i+1,j-1}-D_{i,j-1}}
{x_{i+j}-x_i}
$$

Mediante ciclos iterativos se calculan las diferencias de primer orden hasta el orden cuatro.

```python
columnas = [
    "f[xi]",
    "1as diferencias",
    "2as diferencias",
    "3as diferencias",
    "4as diferencias"
]

tabla = D.copy()

for j in range(1, n):
    tabla[n-j:, j] = np.nan

df = pd.DataFrame(tabla, columns=columnas)

df.insert(0, "xi", x)

display(df.round(7))
```

Calculamos los coeficientes

```python
def polinomio_newton(valor, x, coeficientes):

    resultado = coeficientes[-1]

    for k in range(len(coeficientes) - 2, -1, -1):

        resultado = (
            coeficientes[k]
            + (valor - x[k]) * resultado
        )

    return resultado
```

Y finalmente evaluamos en el punto

```python
# Punto de evaluación
valor = 1.5

# Evaluación del polinomio de Newton
resultado = polinomio_newton(
    valor, x, coeficientes
)

print(f"P_4({valor}) = {resultado:.7f}")
```
